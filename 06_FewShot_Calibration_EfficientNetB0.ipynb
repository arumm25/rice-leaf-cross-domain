# 06_Few-Shot Calibration Menggunakan Feature Extractor EfficientNetB0

Notebook ini mengevaluasi metode few-shot calibration menggunakan EfficientNetB0
sebagai feature extractor pembanding terhadap model baseline MobileNetV2.

Deep feature berukuran 1.280 dimensi dari EfficientNetB0 digunakan sebagai input
untuk klasifikasi menggunakan Linear Support Vector Machine (Linear SVM).

Eksperimen ini menggunakan skenario, pembagian data (master split), dan protokol
evaluasi yang sama dengan notebook MobileNetV2 untuk memastikan perbandingan
performa dilakukan secara adil pada data yang sama.

## Konfigurasi Model

Feature Extractor:
- EfficientNetB0

Classifier:
- Linear SVM

## Skenario Evaluasi

- A → A
- B → B
- A+B → A+B
- A+B → A
- A+B → B
- A → B
- B → A

## Few-Shot Calibration

Evaluasi dilakukan dengan memberikan sejumlah kecil data berlabel dari domain
target untuk mengukur kemampuan adaptasi model terhadap perubahan domain.

Nilai hyperparameter C pada Linear SVM dipilih hanya menggunakan data training
melalui repeated StratifiedGroupKFold, sedangkan data pengujian tidak digunakan
dalam proses pemilihan model untuk mencegah data leakage.

```python
# ============================================================
# CELL 5 — MEMBACA MASTER SPLIT
# ============================================================

SPLIT_NAMES = [
    "train_a", "val_a", "test_a",
    "train_b", "val_b", "test_b",
    "train_ab", "val_ab", "test_ab",
    "external_a_full",
    "external_b_full"
]


LABEL_TO_ID = {
    label: i
    for i, label
    in enumerate(TARGET_CLASSES)
}


splits = {}


for name in SPLIT_NAMES:

    path = (
        SPLIT_DIR
        / f"{name}.csv"
    )

    df = pd.read_csv(path)

    df["label_id"] = (
        df["label"]
        .map(LABEL_TO_ID)
        .astype(int)
    )

    splits[name] = (
        df.reset_index(drop=True)
    )


for name in SPLIT_NAMES:
    print(
        f"{name:<16}: "
        f"{len(splits[name])}"
    )
```

```python
# ============================================================
# CELL 6 — PROTOKOL EKSPERIMEN
# EfficientNetB0 Frozen + SVM
# ============================================================


REGIMES = {

    "a": {
        "train": "train_a",
        "val": "val_a"
    },

    "b": {
        "train": "train_b",
        "val": "val_b"
    },

    "ab": {
        "train": "train_ab",
        "val": "val_ab"
    }

}

# ============================
# MODEL UTAMA PENELITIAN
# ============================

MODEL_NAME = "efficientnetb0_frozen"


MODEL_LABEL = (
    "EfficientNetB0 Frozen"
)

# ============================
# SKENARIO EKSPERIMEN
# ============================

EXPERIMENTS = [

    ("S1", "a", "A_to_A", "test_a"),

    ("S2", "b", "B_to_B", "test_b"),

    ("S3", "a", "A_to_B", "external_b_full"),

    ("S4", "b", "B_to_A", "external_a_full"),

    ("S5", "ab", "AB_to_AB", "test_ab"),

    ("S6", "ab", "AB_to_A", "test_a"),

    ("S7", "ab", "AB_to_B", "test_b")

]

display(
    pd.DataFrame(
        EXPERIMENTS,
        columns=[
            "scenario",
            "regime",
            "evaluation",
            "test_split"
        ]
    )
)
```

```python
# ============================================================
# CELL 7 — HELPER FEATURE
# ============================================================

def feature_dir(model, regime):

    if model == "efficientnetb0_frozen":
        # frozen features flat per-split, tidak ada subfolder regime
        return FROZEN_ROOT

    elif model == "finetuned":
        return FT_ROOT / regime

    else:
        raise ValueError(
            f"Model tidak didukung di notebook ini: {model}"
        )


def load_features(model, regime, split_name):

    directory = feature_dir(model, regime)

    feature_path = directory / f"{split_name}_features.npy"
    label_path = directory / f"{split_name}_labels.npy"
    index_path = directory / f"{split_name}_index.csv"

    if not feature_path.exists():
        raise FileNotFoundError(
            f"Feature tidak ditemukan: {feature_path}"
        )

    X = np.load(feature_path, mmap_mode="r")
    y = np.load(label_path)
    index_df = pd.read_csv(index_path)

    return X, y, index_df


print("Feature loader efficientnetb0_frozen siap.")
```

```python
# ============================================================
# CELL 8 — VALIDASI FEATURE ARTIFACT
# ============================================================

validation_rows = []


# Eksperimen ini hanya menggunakan EfficientNetB0 frozen feature
MODELS_ABLATION = [
    "efficientnetb0_frozen"
]

for model in MODELS_ABLATION:

    for regime, info in REGIMES.items():

        needed = {
            info["train"],
            info["val"]
        }

        needed.update(
            test_split
            for _, r, _, test_split
            in EXPERIMENTS
            if r == regime
        )

        for split_name in sorted(needed):

            X, y, index_df = (
                load_features(
                    model,
                    regime,
                    split_name
                )
            )

            source = splits[
                split_name
            ]


            valid = (

                X.shape
                ==
                (
                    len(source),
                    1280
                )

                and

                np.array_equal(
                    y,
                    source["label_id"]
                    .to_numpy()
                )

                and

                index_df["path"]
                .astype(str)
                .tolist()

                ==

                source["path"]
                .astype(str)
                .tolist()

                and

                np.isfinite(
                    X
                ).all()
            )

            validation_rows.append({

                "model":
                    model,

                "regime":
                    regime,

                "split":
                    split_name,

                "valid":
                    valid
            })

feature_validation = pd.DataFrame(
    validation_rows
)

display(
    feature_validation
)

if not feature_validation[
    "valid"
].all():

    raise RuntimeError(
        "Feature validation gagal."
    )

print(
    "\nFeature validation: LULUS"
)
```

```python
# ============================================================
# CELL 9 — LINEAR SVM HELPER
# ============================================================

def build_svm(C):

    return Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "svm",
            LinearSVC(
                C=C,
                class_weight=CLASS_WEIGHT,
                max_iter=MAX_ITER,
                tol=TOL,
                dual="auto",
                random_state=RANDOM_STATE
            )
        )
    ])


def group_ids(index_df):

    return (
        index_df["dataset"]
        .astype(str)
        +
        "::"
        +
        index_df["split_group_id"]
        .astype(str)
    ).to_numpy()


print("Linear SVM helper siap.")
```

```python
# ============================================================
# CELL 10 — C SELECTION HELPER
# ============================================================

def tune_c(
    model,
    regime
):

    train_split = (
        REGIMES[regime]["train"]
    )

    X, y, index_df = load_features(
        model,
        regime,
        train_split
    )

    groups = group_ids(
        index_df
    )

    rows = []


    for C in C_VALUES:

        scores = []

        for seed in CV_SEEDS:

            cv = StratifiedGroupKFold(
                n_splits=CV_FOLDS,
                shuffle=True,
                random_state=seed
            )

            for fold, (tr, va) in enumerate(
                cv.split(X, y, groups)
            ):

                svm = build_svm(C)

                svm.fit(
                    X[tr],
                    y[tr]
                )

                pred = svm.predict(
                    X[va]
                )

                score = f1_score(
                    y[va],
                    pred,
                    average="macro",
                    zero_division=0
                )

                scores.append(score)


        rows.append({
            "model": model,
            "regime": regime,
            "C": C,
            "mean_macro_f1":
                np.mean(scores),
            "std_macro_f1":
                np.std(scores),
            "n_scores":
                len(scores)
        })


    return pd.DataFrame(rows)
```

```python
# ============================================================
# CELL 11 — RUN C SELECTION (RESUMABLE)
# ============================================================

cv_results = []

ABLATION_MODELS = [
    "efficientnetb0_frozen"
]


for model, regime in product(
    ABLATION_MODELS,
    REGIMES.keys()
):


    output_path = (
        CV_DIR
        / f"{model}_{regime}.csv"
    )


    # Gunakan hasil yang sudah tersedia
    if output_path.exists():

        result = pd.read_csv(
            output_path
        )

        print(
            f"REUSE: {model} / {regime}"
        )


    else:

        print(
            f"\nRUN: {model} / {regime}"
        )


        result = tune_c(
            model,
            regime
        )


        result.to_csv(
            output_path,
            index=False
        )


    cv_results.append(
        result
    )



cv_summary = pd.concat(
    cv_results,
    ignore_index=True
)



CV_SUMMARY_PATH = (
    METRICS_DIR
    /
    "c_selection_summary.csv"
)


cv_summary.to_csv(
    CV_SUMMARY_PATH,
    index=False
)



display(
    cv_summary
)


print(
    "\nC selection selesai."
)
```

```python
# ============================================================
# CELL 12 — BEST C
# ============================================================

best_rows = []


for (model, regime), df in (
    cv_summary.groupby(
        ["model", "regime"]
    )
):

    best = (
        df.sort_values(
            [
                "mean_macro_f1",
                "std_macro_f1",
                "C"
            ],
            ascending=[
                False,
                True,
                True
            ]
        )
        .iloc[0]
    )

    best_rows.append(
        best.to_dict()
    )


best_c = pd.DataFrame(
    best_rows
)


BEST_C_PATH = (
    METRICS_DIR
    / "best_c.csv"
)

best_c.to_csv(
    BEST_C_PATH,
    index=False
)


display(best_c)
```

```python
# ============================================================
# CELL 13 — FINAL TRAINING HELPER
# ============================================================

def train_final_svm(
    model,
    regime,
    C
):

    train_name = (
        REGIMES[regime]["train"]
    )

    val_name = (
        REGIMES[regime]["val"]
    )


    X_train, y_train, _ = (
        load_features(
            model,
            regime,
            train_name
        )
    )

    X_val, y_val, _ = (
        load_features(
            model,
            regime,
            val_name
        )
    )


    X = np.concatenate([
        X_train,
        X_val
    ])

    y = np.concatenate([
        y_train,
        y_val
    ])


    svm = build_svm(C)

    svm.fit(X, y)

    return svm, len(y)
```

```python
# ============================================================
# CELL 14 — TRAIN FINAL SVM
# ============================================================

model_records = []


for _, row in best_c.iterrows():

    model = row["model"]
    regime = row["regime"]
    C = float(row["C"])


    svm, training_rows = (
        train_final_svm(
            model,
            regime,
            C
        )
    )


    model_path = (
        MODEL_DIR
        / f"{model}_{regime}.joblib"
    )


    joblib.dump(
        svm,
        model_path
    )


    model_records.append({
        "model": model,
        "regime": regime,
        "C": C,
        "training_rows":
            training_rows,
        "model_path":
            str(model_path)
    })


final_models = pd.DataFrame(
    model_records
)

display(final_models)
```

```python
# ============================================================
# CELL 15 — EVALUATION HELPER
# ============================================================

def calculate_metrics(
    y_true,
    y_pred
):

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )


    return {
        "accuracy":
            accuracy_score(
                y_true,
                y_pred
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred
            ),

        "macro_precision":
            precision,

        "macro_recall":
            recall,

        "macro_f1":
            f1,

        "weighted_f1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
    }
```

```python
# ============================================================
# CELL 16 — FINAL EVALUATION
# ============================================================

metric_rows = []
prediction_frames = []


EVALUATION_MODELS = [
    "efficientnetb0_frozen"
]


for model in EVALUATION_MODELS:

    for experiment, regime, name, test_split in EXPERIMENTS:


        model_path = (
            MODEL_DIR
            / f"{model}_{regime}.joblib"
        )


        svm = joblib.load(
            model_path
        )


        X, y, index_df = (
            load_features(
                model,
                regime,
                test_split
            )
        )


        start = time.perf_counter()

        pred = svm.predict(
            X
        )

        elapsed = (
            time.perf_counter()
            - start
        )


        metrics = calculate_metrics(
            y,
            pred
        )


        metric_rows.append({

            "experiment":
                experiment,

            "model":
                model,

            "regime":
                regime,

            "evaluation":
                name,

            "test_split":
                test_split,

            "rows":
                len(y),

            **metrics,

            "svm_ms_per_image":
                elapsed / len(y) * 1000
        })


        predictions = (
            index_df[
                [
                    "dataset",
                    "path",
                    "relative_path",
                    "label"
                ]
            ]
            .copy()
        )


        predictions["label_id"] = y

        predictions["prediction_id"] = pred


        predictions["prediction"] = [
            TARGET_CLASSES[i]
            for i in pred
        ]


        predictions["model"] = model

        predictions["regime"] = regime

        predictions["evaluation"] = name


        prediction_frames.append(
            predictions
        )

evaluation_results = pd.DataFrame(
    metric_rows
)


all_predictions = pd.concat(
    prediction_frames,
    ignore_index=True
)

evaluation_results.to_csv(
    METRICS_DIR
    / "evaluation_results.csv",
    index=False
)

all_predictions.to_csv(
    PREDICTION_DIR
    / "all_predictions.csv",
    index=False
)

display(
    evaluation_results
)
```

```python
# ============================================================
# CELL 17 — PER-CLASS METRICS
# ============================================================

class_rows = []


for (
    model,
    regime,
    evaluation
), df in all_predictions.groupby(
    [
        "model",
        "regime",
        "evaluation"
    ]
):

    y_true = df["label_id"].to_numpy()

    y_pred = df["prediction_id"].to_numpy()


    p, r, f, support = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=range(N_CLASSES),
            zero_division=0
        )
    )


    for i, label in enumerate(TARGET_CLASSES):

        class_rows.append({

            "model": model,

            "regime": regime,

            "evaluation": evaluation,

            "class": label,

            "precision": p[i],

            "recall": r[i],

            "f1": f[i],

            "support": support[i]

        })


# ============================================================
# DATAFRAME HASIL
# ============================================================

per_class_metrics = pd.DataFrame(
    class_rows
)


# Urutan skenario untuk jurnal
evaluation_order = [
    "A_to_A",
    "B_to_B",
    "A_to_B",
    "B_to_A",
    "AB_to_AB",
    "AB_to_A",
    "AB_to_B"
]


per_class_metrics["evaluation"] = pd.Categorical(
    per_class_metrics["evaluation"],
    categories=evaluation_order,
    ordered=True
)


per_class_metrics = (
    per_class_metrics
    .sort_values(
        [
            "model",
            "evaluation",
            "class"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# SIMPAN HASIL
# ============================================================

PER_CLASS_PATH = (
    METRICS_DIR /
    "per_class_metrics.csv"
)


per_class_metrics.to_csv(
    PER_CLASS_PATH,
    index=False
)


# ============================================================
# TAMPILKAN SEMUA HASIL
# ============================================================

display(
    per_class_metrics
)


print(
    "\nPer-class metrics tersimpan:"
)

print(
    PER_CLASS_PATH
)
```

```python
# ============================================================
# CELL 18 — BOOTSTRAP MACRO F1 + WILSON ACCURACY CI
# ============================================================

from statsmodels.stats.proportion import proportion_confint

BOOTSTRAP_N = 1000

ci_rows = []


for keys, df in all_predictions.groupby(
    ["model", "regime", "evaluation"]
):

    y_true = df["label_id"].to_numpy()
    y_pred = df["prediction_id"].to_numpy()

    indices = np.arange(len(y_true))
    scores = []

    for i in range(BOOTSTRAP_N):

        sample = resample(
            indices,
            replace=True,
            n_samples=len(indices),
            stratify=y_true,
            random_state=RANDOM_STATE + i
        )

        scores.append(
            f1_score(
                y_true[sample],
                y_pred[sample],
                average="macro",
                zero_division=0
            )
        )


    f1_low, f1_high = np.percentile(
        scores,
        [2.5, 97.5]
    )

    correct = int(
        np.sum(y_true == y_pred)
    )

    acc_low, acc_high = proportion_confint(
        correct,
        len(y_true),
        alpha=0.05,
        method="wilson"
    )


    ci_rows.append({
        "model": keys[0],
        "regime": keys[1],
        "evaluation": keys[2],

        "macro_f1_ci_lower": f1_low,
        "macro_f1_ci_upper": f1_high,

        "macro_f1_ci_degenerate":
            bool(np.isclose(f1_low, f1_high)),

        "accuracy_ci_lower": acc_low,
        "accuracy_ci_upper": acc_high
    })


bootstrap_ci = pd.DataFrame(
    ci_rows
)


bootstrap_ci.to_csv(
    METRICS_DIR
    / "bootstrap_macro_f1_ci.csv",
    index=False
)


display(bootstrap_ci)
```

```python
# ============================================================
# FEW-SHOT CALIBRATION — KONFIGURASI
# ============================================================

FEWSHOT_MODEL = "eeficientNetB0"

# (regime_base, calib_pool_split, eval_split, label)
# regime_base = regime yang dipakai untuk base training (sesuai S3/S4 di EXPERIMENTS)
# calib_pool_split = pool tempat sample target-domain diambil
# eval_split = sama persis dengan test_split baseline S3/S4, biar bisa dibandingkan langsung
FEWSHOT_SCENARIOS = [
    ("a", "train_b", "external_b_full", "A_to_B"),
    ("b", "train_a", "external_a_full", "B_to_A"),
]

K_VALUES = [0, 1, 2, 5, 10, 20]
N_SEEDS = 10
```

```python
# ============================================================
# FEW-SHOT CALIBRATION — FUNGSI INTI
# ============================================================

def sample_k_per_class(X_pool, y_pool, k, rng):
    """Ambil maksimal k sample per kelas dari pool (tanpa replacement)."""
    if k == 0:
        return (
            np.empty((0, X_pool.shape[1])),
            np.empty((0,), dtype=y_pool.dtype),
        )

    idx_selected = []
    for cls in np.unique(y_pool):
        cls_idx = np.where(y_pool == cls)[0]
        n_take = min(k, len(cls_idx))
        chosen = rng.choice(cls_idx, size=n_take, replace=False)
        idx_selected.extend(chosen.tolist())

    idx_selected = np.array(idx_selected, dtype=int)
    return X_pool[idx_selected], y_pool[idx_selected]


def run_fewshot_calibration(
    model, regime_base, calib_pool_split, eval_split,
    k_values, n_seeds, scenario_label
):
    # Base training = train + val regime (sama seperti train_final_svm di notebook 05)
    train_name = REGIMES[regime_base]["train"]
    val_name = REGIMES[regime_base]["val"]

    X_train, y_train, _ = load_features(model, regime_base, train_name)
    X_val, y_val, _ = load_features(model, regime_base, val_name)

    X_base = np.concatenate([np.asarray(X_train), np.asarray(X_val)])
    y_base = np.concatenate([y_train, y_val])

    # Pool kalibrasi dari domain target (TIDAK overlap dengan eval_split)
    X_pool, y_pool, _ = load_features(model, regime_base, calib_pool_split)
    X_pool = np.asarray(X_pool)

    # Eval set — sama persis dengan test_split baseline S3/S4
    X_eval, y_eval, _ = load_features(model, regime_base, eval_split)
    X_eval = np.asarray(X_eval)

    # Ambil C yang sama dengan yang dipakai baseline S3/S4 untuk regime ini
    C = float(
        best_c.loc[
            (best_c["model"] == model) & (best_c["regime"] == regime_base),
            "C"
        ].iloc[0]
    )

    results = []
    for k in k_values:
        seeds = range(n_seeds) if k > 0 else [0]
        for seed in seeds:
            rng = np.random.default_rng(seed)
            X_calib, y_calib = sample_k_per_class(X_pool, y_pool, k, rng)

            X_fit = np.vstack([X_base, X_calib]) if k > 0 else X_base
            y_fit = np.concatenate([y_base, y_calib]) if k > 0 else y_base

            svm = build_svm(C)
            svm.fit(X_fit, y_fit)

            y_pred = svm.predict(X_eval)
            macro_f1 = f1_score(y_eval, y_pred, average="macro", zero_division=0)
            acc = accuracy_score(y_eval, y_pred)

            results.append({
                "scenario": scenario_label,
                "model": model,
                "regime_base": regime_base,
                "k": k,
                "seed": seed,
                "n_calib_used": len(y_calib),
                "C": C,
                "macro_f1": macro_f1,
                "accuracy": acc,
            })

    return pd.DataFrame(results)


print("Fungsi few-shot calibration siap.")
```

```python
# ============================================================
# FEW-SHOT CALIBRATION — JALANKAN (FIXED)
# ============================================================

# Memperbaiki typo pada nama model agar sesuai dengan model yang didukung
FEWSHOT_MODEL = "efficientnetb0_frozen"

all_fewshot_results = []

for regime_base, calib_pool_split, eval_split, label in FEWSHOT_SCENARIOS:
    print(f"Running few-shot calibration: {label} (model={FEWSHOT_MODEL})")
    df_result = run_fewshot_calibration(
        model=FEWSHOT_MODEL,
        regime_base=regime_base,
        calib_pool_split=calib_pool_split,
        eval_split=eval_split,
        k_values=K_VALUES,
        n_seeds=N_SEEDS,
        scenario_label=label,
    )
    all_fewshot_results.append(df_result)

fewshot_raw_df = pd.concat(all_fewshot_results, ignore_index=True)
display(fewshot_raw_df.head(20))
```

```python
fewshot_summary_df = (
    fewshot_raw_df
    .groupby(["scenario", "model", "k"])
    .agg(
        macro_f1_mean=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),
        accuracy_mean=("accuracy", "mean"),
        accuracy_std=("accuracy", "std"),
        n_calib_used_mean=("n_calib_used", "mean"),
    )
    .reset_index()
    .sort_values(["scenario", "k"])
)

display(fewshot_summary_df)
```
